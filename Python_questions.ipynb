```python
# Palindrome or not
def palindrome(input_val):
    rever_str = input_val[::-1].lower()
    if rever_str==input_val.lower():
        print("Palindrome")
    else:
        print("Not Palindrome")
    return None
```

```python
palindrome('Naman')
palindrome('Namar')
```

```python
# PRime
def isprime(n):
    for i in range(2,int(n**0.5)+1):
        print(i)
        if n%i == 0:
            return False
    return True
```

```python
def fibonacci(n):
    sequence = []
    a, b = 0, 1
    for _ in range(n):
        sequence.append(a)
        a, b = b, a + b
    return sequence
fibonacci(12)
```

```python
from collections import Counter

def letter_frequency(s):
    return Counter(s)

# Example
text = "aravind"
freq = letter_frequency(text)
print(freq)
# Output: Counter({'a': 2, 'r': 1, 'v': 1, 'i': 1, 'n': 1, 'd': 1})
```

```python
str1 = 'gaddam aravind'
freq = {}
for i in str1:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
print(freq)
```

```python
most_common = max(freq, key=freq.get)
most_common
```

```python
def are_anagrams(str1, str2):
    return sorted(str1) == sorted(str2)
are_anagrams('listen', 'silent')
```

```python
def count_vowels(s):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in s:
        if ch in vowels:
            count += 1
    return count

# Example
print(count_vowels("aravind"))   # Output: 3 (a, a, i)
```

```python
s = "Find the longest word in this sentence"
length = {}
for val in s.split():
    leng = len(val)
    length[val] = leng
length
max_word = max(length , key = length.get)
max_word
```
